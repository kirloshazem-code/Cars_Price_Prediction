```python
z = df[['horsepower', 'curb-weight', 'engine-size', 'highway-mpg']]
```

```python
lm.fit(z, df['price'])
```

```python
y_hat = lm.predict(z)
y_hat
```

### Evaluation Model Using Visualization

```python
plt.figure(figsize=(12,6))

ax1 = sns.distplot(df['price'], hist=False, color='r', label='Actual Price')
sns.distplot(y_hat, hist=False, color='b', label='predicted price', ax=ax1)
plt.title('actual VS predicted prices')
plt.ylabel('proportions of cars')
plt.xlabel('price')

plt.show()
```
